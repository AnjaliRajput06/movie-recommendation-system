```python
movies.dropna(inplace = True)
movies.duplicated().sum()
```

```python
import ast  #Abstract Syntax Tree

def convert(obj):
    L = []
    try:
        data = ast.literal_eval(obj)   
        for item in data:
            L.append(item['name'])
    except:
        return []
    return L
```

```python
movies['genres'] = movies['genres'].apply(convert)
print(movies['genres'].head())
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
movies.head()
```

```python
movies['overview'][0]
```

```python
movies['overview'] = movies['overview'].apply(lambda x:x.split())
movies
```

```python
movies['genres'] = movies['genres'].apply(lambda x:[i.replace(" ","") for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
movies.head()
```

```python
movies['tagline'] = movies['overview']+movies['genres']+movies['keywords']
movies
```

```python
new_df = movies[['id','title','tagline']]
new_df
```

```python
new_df['tagline'] = new_df['tagline'].apply(lambda x:' '.join(x))
new_df
```

```python
new_df['tagline'][0]
```

```python
new_df['tagline'] = new_df['tagline'].apply(lambda X:X.lower())
new_df
```

```python
from sklearn.feature_extraction.text import CountVectorizer   # convert text to vectors
cv = CountVectorizer(max_features = 5000, stop_words = 'english')   # keep top 5000 words, remove stopwords
```

```python
cv.fit_transform(new_df['tagline']).toarray().shape # show number of rows and columns
```

```python
vectors = cv.fit_transform(new_df['tagline']).toarray()
vectors[0]
```

```python
len(cv.get_feature_names_out())
```

```python
import nltk  
from nltk.stem.porter import PorterStemmer  
ps = PorterStemmer()
```

```python
def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
        return " ".join(y)
```

```python
new_df['tagline'] = new_df['tagline'].apply(stem)
```

```python
from sklearn.metrics.pairwise import cosine_similarity    # convert text to vectors
cosine_similarity(vectors)
```

```python
cosine_similarity(vectors).shape
```

```python
similarity = cosine_similarity(vectors)
similarity[0]
```

```python
similarity[0].shape
```

```python
sorted(list(enumerate(similarity[0])), reverse = True, key = lambda x:x[1])[1:6]
```

```python
def recommend(movie):  # function to recommend movies
    movie_index = new_df[new_df['title'] == movie].index[0]
    distances = similarity[movie_index]
    movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])[1:6]

    for i in movies_list:
        print(new_df.iloc[i[0]].title)  # select rows/columns by index
```

```python
recommend('Titanic')
```
